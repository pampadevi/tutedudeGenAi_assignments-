TASK 3 :User Menu

```python
# use while break and keep order amount input untill anyone press quit
```

```python

tax_rate = 0.05 # Using tax_rate to avoid confusion with tax amount
running_order_amount=0
running_total=0
while True:
    order_input = input("Enter order amount (or 'q' to quit): ")

    if order_input.lower() == 'q':
        print("Exiting order processing.")
        break

    try:
        order_amount = int(order_input)

        if order_amount > 2000:
            discount = 15
        elif order_amount < 2000 and order_amount >= 1500:
            discount = 10
        elif order_amount < 1500 and order_amount >= 1000:
            discount = 7
        else:
            discount = 0

        subtotal = (100 - discount) * order_amount / 100
        #tax_amount = tax_rate * subtotal
        running_order_amount=order_amount+running_order_amount
        running_total = running_total+subtotal

        print("USER MENU")
        print("Order amount entered:",order_amount,
        "Running order amount:",running_order_amount,
        "Runnig total:",running_total)


    except ValueError:
        print("Invalid input. Please enter a number or 'q' to quit.")
        continue
```
